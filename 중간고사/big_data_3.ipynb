###파일 읽기 함수
read(): 파일을 전체 읽어서 문자열로 반환
readline(): 한줄만 읽기
readlines(): 파일을 읽어서 리스트로 반환

```python
f=open("intro.txt","r")
data=f.read()
print(data)
f.close()
```

```python
f=open("intro.txt","r")
data=f.read()
print(data)
f.close()
```

```python
f=open("intro.txt","r")
s=f.readline()
print(s)
f.close()
```

```python
f=open("intro.txt","r")
s=f.readlines()
print(s)
f.close()
```

```python
f=open("intro.txt","r")
data=f.readlines()
for d in data:
    print(d,end="")
f.close()
```

```python
f=open("intro.txt","a")
f.write("집가고싶은 씹학년\n")
f.close()
```

```python
f=open("intro.txt","r")
data=f.readlines()
for d in data:
    print(d,end="")
f.close()
```

```python
with open("number.txt","w") as f:
    f.write("136846352번째 응가\n")
```

```python
with open("number.txt","r") as f:
    data=f.read()
    print(data)
```
