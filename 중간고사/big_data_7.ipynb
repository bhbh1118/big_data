# 공공데이터포털(www.data.go.kr)접속
## 경기도 의정부시와 대전교통공사,서울교통공사 경전철 및 지하철 혼잡도 데이터 csv파일

```python
# 먼저 빈 워크북을 만들어 wb변수에 담고(처음 생성하면 기본 워크시트 하나)
# 활성화된 워크시트를 ws라고 저장
import openpyxl

wb = openpyxl.Workbook()
ws = wb.active
```

```python
import csv

with open('data/data1.csv', encoding = 'cp949') as f:
    reader = csv.reader(f)
    for row in reader:
        ws.append(row)
```

```python
ws2 = wb.create_sheet('Sheet2')

with open('data/data2.csv', encoding = 'cp949') as f:
    reader = csv.reader(f)
    for row in reader:
        ws2.append(row)
```

```python
ws3 = wb.create_sheet('Sheet3')

with open('data/data3.csv', encoding = 'cp949') as f:
    reader = csv.reader(f)
    for row in reader:
        ws3.append(row)
```

```python
ws.title = '경기도 의정부시_의정부경전철 혼잡도'
ws2.title = '대전교통공사_열차 혼잡도 분석'
ws3.title = '서울특별시_지하철 혼잡도 정보'

wb.save('data/traffic.xlsx')
```
