# Q1. 자동차 배기량에 따라 고속도로 연비가 다른지 알아볼라 함
- displ(배기량)이 4이하인 차랑 5이상인 차 중 어떤 차의 hwy(고속도로 연비) 평균이 더 높은지 알아봐라

```python
import pandas as pd

mpg = pd.read_csv('data/mpg.csv')

mpg_a = mpg.query('displ <= 4')
mpg_b = mpg.query('displ >= 5')
```

```python
mpg_a
```

```python
# 배기량이 4 이하의 평균
mpg_a['hwy'].mean()
```

```python
# 배기량이 5 이상 평균
mpg_b['hwy'].mean()
```

# Q2. 아우디와 도요다중 어느 manufacturer의 cty(도시 연비) 평균
- 뭐가 더 높노??

```python
mpg_audi = mpg.query('manufacturer == "audi"')
mpg_toyota = mpg.query('manufacturer == "toyota"')
```

```python
# audi 연비 평균
mpg_audi['cty'].mean()
```

```python
# toyota 연비 평균
mpg_toyota['cty'].mean()
```
