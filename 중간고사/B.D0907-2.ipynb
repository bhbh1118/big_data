```python
import openpyxl

wb = openpyxl.load_workbook('data/characters.xlsx')
```

```python
print(wb.sheetnames)
```

```python
#새 워크북과 워크시트 생성
wb.create_sheet('Sheet2')
print(wb.sheetnames)
```

```python
#워크시트 이름 변경
GAYBAR = wb['GAYBAR']
TRASGENDER = wb['TRASGENDER']

GAYBAR.title = '캐릭터 명단'
TRASGENDER.title = '인기도 조사'

print(wb.sheetnames)
```

```python
sheet2['B1'] = '인기도 조사결과'
print(sheet2['B1'].value)
```

```python
copysheet = wb.copy_worksheet(sheet2)
print(wb.sheetnames)

#복제 시트의 이름 변경하기
copysheet.title = 'copy'

print(wb.sheetnames)
```
